## The hypotheses that were tested

```python
from bionexus.adapters.store.sqlite import SqliteProvenanceStore
store = SqliteProvenanceStore("campaign.db")
for r in store.branch_records("campaign-root"):
    if r.event_type.value.startswith("hypothesis"):
        print(f"{r.event_type.value:24} {(r.artifact.metadata or {}).get('statement','')[:70]}")
```
