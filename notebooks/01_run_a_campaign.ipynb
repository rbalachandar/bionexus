Every line above is a provenance-recorded event. Inspect the hash chain:

```python
import subprocess, sys
print(subprocess.run([sys.executable, '-m', 'bionexus.cli', 'prove', 'campaign-root', '--db', 'campaign.db'], capture_output=True, text=True).stdout)
```

## The hypotheses that were tested

Each hypothesis follows the lifecycle `proposed -> tested -> validated/invalidated`, with evidence recorded.

```python
from bionexus.adapters.store.sqlite import SqliteProvenanceStore
store = SqliteProvenanceStore('campaign.db')
for r in store.branch_records('campaign-root'):
    if r.event_type.value.startswith('hypothesis'):
        print(f"{r.event_type.value:24} {(r.artifact.metadata or {}).get('statement','')[:70]}")
```

## Shareable report

Render the run as a self-contained HTML report (chain verification + hypothesis stats included):

```python
subprocess.run([sys.executable, '-m', 'bionexus.cli', 'prove', 'campaign-root', '--db', 'campaign.db', '--html', 'campaign_report.html'])
print('open campaign_report.html in a browser')
```
